```python

%%writefile /content/YouTube_BigData_CA2/dashboard_app.py

import os
import glob
import pandas as pd
import numpy as np
import streamlit as st
import plotly.express as px
import plotly.graph_objects as go

# --------------------------------------------------
# PAGE CONFIGURATION
# --------------------------------------------------

st.set_page_config(
    page_title="YouTube Competitive Intelligence",
    page_icon="▶",
    layout="wide"
)

PROJECT_DIR = "/content/YouTube_BigData_CA2"

st.title("YouTube Channel Competitive Intelligence")
st.caption("Big Data Analytics | Competitor Benchmarking and Content Strategy")

# --------------------------------------------------
# DATA LOADING
# --------------------------------------------------

def find_csv(pattern):
    matches = glob.glob(
        os.path.join(PROJECT_DIR, "**", pattern),
        recursive=True
    )
    if not matches:
        raise FileNotFoundError(f"Could not find: {pattern}")
    return matches[0]


@st.cache_data
def load_data():
    data = {}

    data["summary"] = pd.read_csv(
        find_csv("project_summary.csv")
    )
    data["strategies"] = pd.read_csv(
        find_csv("strategy_recommendations.csv")
    )
    data["channels"] = pd.read_csv(
        find_csv("competitor_channels.csv")
    )
    data["channel_info"] = pd.read_csv(
        find_csv("channel_data.csv")
    )
    data["videos"] = pd.read_csv(
        find_csv("all_videos_cleaned.csv")
    )
    data["sentiment"] = pd.read_csv(
        find_csv("comment_sentiment.csv")
    )
    data["clusters"] = pd.read_csv(
        find_csv("canopy_clusters.csv")
    )
    data["topics"] = pd.read_csv(
        find_csv("part-00000-fc8400e4-7f1a-4c0b-a0d3-3adeda629e9a-c000.csv")
    )
    data["publishing"] = pd.read_csv(
        find_csv("part-00000-36b6547b-6cc1-48bb-af51-d0a21784073f-c000.csv")
    )
    data["gaps"] = pd.read_csv(
        find_csv("part-00000-bd273bb3-bef5-4527-9342-0ee96ad9d4cc-c000.csv")
    )
    data["performance"] = pd.read_csv(
        find_csv("part-00000-26c5ad5d-2bb4-4471-982a-ffa5bc1eb581-c000.csv")
    )

    return data


try:
    data = load_data()
except Exception as e:
    st.error(f"Unable to load project data: {e}")
    st.stop()

summary = data["summary"]
videos = data["videos"].copy()
performance = data["performance"].copy()
sentiment = data["sentiment"].copy()
clusters = data["clusters"].copy()
topics = data["topics"].copy()
gaps = data["gaps"].copy()
publishing = data["publishing"].copy()
strategies = data["strategies"].copy()

# Convert numerical columns safely
numeric_columns = [
    "views", "likes", "comments_count",
    "engagement_rate", "publish_year"
]

for col in numeric_columns:
    if col in videos.columns:
        videos[col] = pd.to_numeric(videos[col], errors="coerce")

for col in [
    "average_views", "average_likes",
    "average_comments", "avg_engagement_rate",
    "total_sample_views", "videos_analyzed"
]:
    if col in performance.columns:
        performance[col] = pd.to_numeric(
            performance[col], errors="coerce"
        )

# --------------------------------------------------
# SIDEBAR FILTERS
# --------------------------------------------------

st.sidebar.header("Dashboard Filters")

channels = sorted(videos["channel_name"].dropna().unique())

selected_channels = st.sidebar.multiselect(
    "Select competitor channels",
    channels,
    default=channels
)

filtered_videos = videos[
    videos["channel_name"].isin(selected_channels)
].copy()

filtered_performance = performance[
    performance["channel_name"].isin(selected_channels)
].copy()

st.sidebar.caption(
    f"{len(selected_channels)} of {len(channels)} channels selected"
)

# --------------------------------------------------
# KPI CARDS
# --------------------------------------------------

def summary_value(metric):
    result = summary.loc[
        summary["metric"] == metric, "value"
    ]
    return float(result.iloc[0]) if not result.empty else 0


total_channels = int(summary_value("Competitor channels discovered"))
total_videos = int(summary_value("Competitor videos collected"))
total_comments = int(summary_value("Viewer comments collected"))
total_clusters = int(summary_value("Canopy clusters generated"))

total_views = (
    filtered_videos["views"].sum()
    if not filtered_videos.empty else 0
)

st.subheader("Project Overview")

c1, c2, c3, c4 = st.columns(4)

c1.metric("Competitor Channels", f"{total_channels:,}")
c2.metric("Videos Analyzed", f"{total_videos:,}")
c3.metric("Comments Analyzed", f"{total_comments:,}")
c4.metric("Canopy Clusters", f"{total_clusters:,}")

st.info(
    "Scope: 29 discovered competitor channels and 816 collected competitor videos. "
    "No target-channel video records were collected, so this dashboard focuses "
    "on competitor benchmarking and content opportunities."
)

# --------------------------------------------------
# TABS
# --------------------------------------------------

tab1, tab2, tab3, tab4, tab5, tab6 = st.tabs([
    "Competitor Analysis",
    "Engagement",
    "Publishing Trends",
    "Content Intelligence",
    "Audience Sentiment",
    "Canopy Clustering"
])

# --------------------------------------------------
# TAB 1: COMPETITOR ANALYSIS
# --------------------------------------------------

with tab1:
    st.subheader("Competitor Performance")

    if not filtered_performance.empty:
        ranking = filtered_performance.sort_values(
            "average_views", ascending=False
        ).head(15)

        fig = px.bar(
            ranking.sort_values("average_views"),
            x="average_views",
            y="channel_name",
            orientation="h",
            color="average_views",
            title="Top Competitors by Average Video Views",
            labels={
                "average_views": "Average Views",
                "channel_name": "Channel"
            },
            color_continuous_scale="Blues"
        )
        fig.update_layout(height=550)
        st.plotly_chart(fig, use_container_width=True)

        st.subheader("Competitor Ranking Table")
        st.dataframe(
            ranking.sort_values(
                "average_views", ascending=False
            ).reset_index(drop=True),
            use_container_width=True,
            hide_index=True
        )
    else:
        st.warning("No competitor data available for this selection.")

# --------------------------------------------------
# TAB 2: ENGAGEMENT
# --------------------------------------------------

with tab2:
    st.subheader("Competitor Engagement Analysis")

    if not filtered_performance.empty:
        fig = px.scatter(
            filtered_performance,
            x="average_views",
            y="avg_engagement_rate",
            size="videos_analyzed",
            color="channel_name",
            hover_name="channel_name",
            title="Average Views vs Engagement Rate",
            labels={
                "average_views": "Average Views",
                "avg_engagement_rate": "Average Engagement Rate (%)",
                "videos_analyzed": "Videos Analyzed"
            }
        )
        st.plotly_chart(fig, use_container_width=True)

        engagement_ranking = filtered_performance.sort_values(
            "avg_engagement_rate", ascending=False
        )

        fig2 = px.bar(
            engagement_ranking,
            x="channel_name",
            y="avg_engagement_rate",
            title="Average Engagement Rate by Channel",
            labels={
                "channel_name": "Channel",
                "avg_engagement_rate": "Engagement Rate (%)"
            },
            color="avg_engagement_rate",
            color_continuous_scale="Viridis"
        )
        fig2.update_layout(xaxis_tickangle=-45)
        st.plotly_chart(fig2, use_container_width=True)

    if not filtered_videos.empty:
        st.subheader("Video-Level Performance")
        st.dataframe(
            filtered_videos[
                ["channel_name", "title", "views",
                 "likes", "comments_count", "engagement_rate"]
            ].sort_values("views", ascending=False).head(20),
            use_container_width=True,
            hide_index=True
        )

# --------------------------------------------------
# TAB 3: PUBLISHING TRENDS
# --------------------------------------------------

with tab3:
    st.subheader("Competitor Publishing Frequency")

    publishing_filtered = publishing[
        publishing["channel_name"].isin(selected_channels)
    ].copy()

    publishing_filtered["avg_monthly_videos"] = pd.to_numeric(
        publishing_filtered["avg_monthly_videos"],
        errors="coerce"
    )

    fig = px.bar(
        publishing_filtered.sort_values("avg_monthly_videos"),
        x="avg_monthly_videos",
        y="channel_name",
        orientation="h",
        color="avg_monthly_videos",
        title="Average Videos Published per Active Month",
        labels={
            "avg_monthly_videos": "Average Monthly Videos",
            "channel_name": "Channel"
        },
        color_continuous_scale="Teal"
    )
    fig.update_layout(height=600)
    st.plotly_chart(fig, use_container_width=True)

    st.dataframe(
        publishing_filtered.sort_values(
            "avg_monthly_videos", ascending=False
        ),
        use_container_width=True,
        hide_index=True
    )

# --------------------------------------------------
# TAB 4: CONTENT INTELLIGENCE
# --------------------------------------------------

with tab4:
    st.subheader("Popular Topics")

    top_topics = topics.sort_values(
        "frequency", ascending=False
    ).head(20)

    fig = px.bar(
        top_topics.sort_values("frequency"),
        x="frequency",
        y="word",
        orientation="h",
        color="frequency",
        title="Most Frequently Occurring Topic Keywords",
        labels={
            "frequency": "Frequency",
            "word": "Keyword"
        },
        color_continuous_scale="Blues"
    )
    fig.update_layout(height=600)
    st.plotly_chart(fig, use_container_width=True)

    st.subheader("Competitor-Derived Content Opportunities")

    top_gaps = gaps.sort_values(
        "competitor_gap", ascending=False
    ).head(20)

    fig2 = px.bar(
        top_gaps.sort_values("competitor_gap"),
        x="competitor_gap",
        y="word",
        orientation="h",
        color="competitor_gap",
        title="Top Content Opportunity Keywords",
        labels={
            "competitor_gap": "Competitor Share Difference",
            "word": "Keyword"
        },
        color_continuous_scale="Viridis"
    )
    fig2.update_layout(height=600)
    st.plotly_chart(fig2, use_container_width=True)

    st.caption(
        "These opportunities reflect the difference between competitor "
        "keyword share and target-channel keyword share. Target-channel "
        "video data was unavailable, so target share is zero in this analysis."
    )

    st.subheader("Rule-Based Strategy Recommendations")
    for _, row in strategies.iterrows():
        with st.expander(str(row["strategy_category"])):
            st.write(row["recommendation"])
            st.caption(f"Evidence: {row['evidence']}")

# --------------------------------------------------
# TAB 5: AUDIENCE SENTIMENT
# --------------------------------------------------

with tab5:
    st.subheader("Comment Sentiment Analysis")

    sentiment_counts = (
        sentiment["sentiment"]
        .value_counts()
        .reindex(["Positive", "Neutral", "Negative"], fill_value=0)
        .reset_index()
    )
    sentiment_counts.columns = ["Sentiment", "Comments"]

    s1, s2, s3 = st.columns(3)
    counts = dict(zip(
        sentiment_counts["Sentiment"],
        sentiment_counts["Comments"]
    ))

    s1.metric("Positive Comments", f"{counts['Positive']:,}")
    s2.metric("Neutral Comments", f"{counts['Neutral']:,}")
    s3.metric("Negative Comments", f"{counts['Negative']:,}")

    fig = px.pie(
        sentiment_counts,
        names="Sentiment",
        values="Comments",
        hole=0.45,
        title="Distribution of Viewer Comment Sentiment"
    )
    st.plotly_chart(fig, use_container_width=True)

    fig2 = px.bar(
        sentiment_counts,
        x="Sentiment",
        y="Comments",
        color="Sentiment",
        title="Comment Sentiment Counts"
    )
    st.plotly_chart(fig2, use_container_width=True)

    st.caption(
        "Sentiment was calculated using VADER on the 394 collected comments."
    )

# --------------------------------------------------
# TAB 6: CANOPY CLUSTERING
# --------------------------------------------------

with tab6:
    st.subheader("Canopy Clustering Analysis")

    cluster_summary = (
        clusters.groupby("canopy_cluster")
        .agg(
            video_count=("video_id", "count"),
            average_views=("views", "mean"),
            average_engagement=("engagement_rate", "mean")
        )
        .reset_index()
        .sort_values("video_count", ascending=False)
    )

    fig = px.bar(
        cluster_summary.sort_values("canopy_cluster"),
        x="canopy_cluster",
        y="video_count",
        color="video_count",
        title="Number of Videos in Each Canopy Cluster",
        labels={
            "canopy_cluster": "Cluster",
            "video_count": "Number of Videos"
        },
        color_continuous_scale="Blues"
    )
    st.plotly_chart(fig, use_container_width=True)

    fig2 = px.scatter(
        cluster_summary,
        x="average_views",
        y="average_engagement",
        size="video_count",
        color="canopy_cluster",
        hover_data=["video_count"],
        title="Cluster Performance: Average Views vs Engagement",
        labels={
            "average_views": "Average Views",
            "average_engagement": "Average Engagement Rate (%)",
            "canopy_cluster": "Cluster"
        }
    )
    st.plotly_chart(fig2, use_container_width=True)

    st.subheader("Cluster Statistics")
    st.dataframe(
        cluster_summary.round(2),
        use_container_width=True,
        hide_index=True
    )

    st.caption(
        "Canopy clustering grouped videos using the project's selected "
        "performance features. Cluster IDs are algorithm-generated labels, "
        "not predefined content categories."
    )

# --------------------------------------------------
# FOOTER
# --------------------------------------------------

st.divider()
st.caption(
    "YouTube Competitive Intelligence Dashboard | "
    "BE Computer Engineering CA2 | Built using Streamlit and Plotly"
)
```
